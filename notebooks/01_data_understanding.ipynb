# Online Retail Data Analysis

## 1. Business Context

This project analyzes transaction-level data from an online retail business.

The objective is to understand sales performance, customer purchasing behavior,
product performance, and geographic trends using Python, Pandas, NumPy, and
data visualization.

### Key Business Questions

- How much revenue does the business generate?
- How does revenue change over time?
- Which products generate the most revenue?
- Which customers contribute the most revenue?
- Which countries generate the most sales?
- How frequently do customers purchase?
- Are there cancelled or unusual transactions?
- What patterns can help the business improve sales and customer retention?
